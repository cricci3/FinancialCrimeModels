```python
from workflow.internal import load_dataset, knn_graph, normalization, visualize_metrics
from workflow.SQUIC_functions import squic_fit_matrix_computation, squic_fit_computation
from workflow.clustering_functions import clustering_2_communities, modularity_fscore, plot_Q_f1
```

```python
# Load dataset (the user will pass the name)
Y, name, dimension, account_prop, trans_matrix = load_dataset()


# Normalise time series
Y_norm = normalization(Y, name)

results_squic = {}

# Run SQUIC_fit
results_squic['squic-fit-matrix'] = squic_fit_computation(Y_norm, name, dimension, printMatrix=True)
```

```python
import numpy as np
import networkx as nx
```

```python
for technique, matrices in results_squic.items():
        for rho, X in matrices.items():
            X = np.abs(X) 

            # Create a graph from matrix X
            G = nx.from_scipy_sparse_array(X)

            print(f"for rho {rho} graph is connected {nx.is_connected(G)}")
```

```python
dict_cluster = clustering_2_communities(results_squic, name, account_prop)

metrics = modularity_fscore(dict_cluster, results_squic, account_prop)

plot_Q_f1(metrics)
```
