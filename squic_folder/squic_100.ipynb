```python
DATASET = "../datasets/AMLSimData/100"

accounts = pd.read_csv(f'{DATASET}/accounts.csv')
transactions = pd.read_csv(f'{DATASET}/transactions.csv')
```

```python
# Create root
balances = {}

# Take users data from account.csv
for _, row in accounts.iterrows():
    acc_id = row["ACCOUNT_ID"] # User id
    open_date = 0
    initial_balance = round(float(row["INIT_BALANCE"]), 2)

    balances[acc_id] = [{"date": open_date,
                         "balance": initial_balance
                        }]
    
    if acc_id == 0: # Check
        print("first user must be 0 => ",str(acc_id))
        print("user 0 initial deposit must be 184.44 => ", initial_balance)
```

```python
# Apply transactions to users' balances
transactions.sort_values(by="TX_ID")

# Track the last balance for each account
current_balances = {acct_id: balances[acct_id][0]["balance"] for acct_id in balances}

for i, (_, row) in enumerate(transactions.iterrows()):
    orig_acct = row["SENDER_ACCOUNT_ID"]
    bene_acct = row["RECEIVER_ACCOUNT_ID"]
    amount = row["TX_AMOUNT"]
    tx_type = row['TX_TYPE']
    date = row["TIMESTAMP"]

    # if i == 0:
    #     print("first transaction must be from 6456 to 9069 => ", orig_acct, bene_acct)
    # if i == 1323233: # check that read till the last transaction
    #     print("\nlast transaction must be from 545 to 2724 => ", orig_acct, bene_acct)

    if orig_acct in balances: # check that sender has a deposit
        last_balance = balances[orig_acct][-1]["balance"] # read balance from balances
        if tx_type in ['TRANSFER', 'WITHDRAWAL']: # all TRANSFER in the csv
            new_balance = last_balance - amount
        # elif tx_type == 'DEPOSIT':
        #    new_balance = last_balance + amount

        balances[orig_acct].append({"date": date, "balance": round(new_balance, 2)})
        # if orig_acct == 1 and date == 1:
        #     print()
        #     print(row['TX_ID'])
        #     print(f"on date {date} user {orig_acct} send to {bene_acct} {amount} dollars")
        #     print(f"user {orig_acct} balance goes from {last_balance} to {new_balance}")
        #     print(balances[orig_acct])
            
    
    if bene_acct in balances: # check that receiver has a deposit
        last_balance = balances[bene_acct][-1]["balance"]
        if tx_type in ['TRANSFER', 'DEPOSIT']:
            new_balance = last_balance + amount
        # elif tx_type == 'WITHDRAWAL':
        #     new_balance = last_balance - amount

        balances[bene_acct].append({"date": date, "balance": round(new_balance, 2)})
        # if bene_acct == 0:
        #     print()
        #     print(row['TX_ID'])
        #     print(f"on date {date} user {orig_acct} send to {bene_acct} {amount} dollars")
        #     print(f"user {bene_acct} balance goes from {last_balance} to {new_balance}")
        #     print(balances[bene_acct])
            
```

```python
from sklearn.preprocessing import MinMaxScaler
import networkx as nx
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import SpectralClustering, AffinityPropagation, DBSCAN
from sklearn.metrics import silhouette_score
from networkx.algorithms import community

def cluster_and_evaluate(X, method='spectral', param_range=range(2,11)):
    sil_scores = []
    valid_params = []
    
    for param in param_range:
        try:
            if method == 'spectral':
                clustering = SpectralClustering(n_clusters=param, 
                                             affinity='precomputed',
                                             assign_labels='discretize',
                                             random_state=42)
                labels = clustering.fit_predict(np.asarray(X))
            elif method == 'affinity':
                clustering = AffinityPropagation(affinity='precomputed',
                                               damping=0.5 + param*0.05,
                                               random_state=42)
                labels = clustering.fit_predict(np.asarray(X))
            elif method == 'dbscan':
                clustering = DBSCAN(eps=param*0.1, 
                                 min_samples=5,
                                 metric='precomputed')
                labels = clustering.fit_predict(np.asarray(X))
                
            # Only calculate silhouette if we have more than 1 cluster and less than n_samples/2
            unique_labels = np.unique(labels)
            if len(unique_labels) > 1 and len(unique_labels) < len(X)/2:
                score = silhouette_score(np.asarray(X), labels, metric='precomputed')
                sil_scores.append(score)
                valid_params.append(param)
            else:
                sil_scores.append(np.nan)
        except Exception as e:
            print(f"Error with {method} (param={param}): {str(e)}")
            sil_scores.append(np.nan)
    
    return valid_params, sil_scores

def louvain_clustering(X):
    G = nx.from_numpy_array(X)
    partition = community.louvain_communities(G, weight='weight', resolution=1.0)
    labels = np.zeros(len(X), dtype=int)
    for i, community_nodes in enumerate(partition):
        for node in community_nodes:
            labels[node] = i
    return labels

for rho in l:
    X = fit_norm_dict[rho]

    np.fill_diagonal(X, 0)  # Ensure diagonal is zero
    X = np.abs(X)  # Ensure all values are positive
    
    # Method 1: Spectral clustering
    spectral_params, spectral_scores = cluster_and_evaluate(X, 'spectral')
    
    # Method 2: Affinity propagation
    aff_params = np.linspace(0.5, 0.9, 8)
    _, aff_scores = cluster_and_evaluate(X, 'affinity', aff_params)
    
    # Method 3: Louvain
    louvain_labels = louvain_clustering(X)
    louvain_score = silhouette_score(np.asarray(X), louvain_labels, metric='precomputed')
    
    # Plot results
    plt.figure(figsize=(12, 5))
    plt.subplot(1, 2, 1)
    
    # Only plot valid scores
    if len(spectral_params) > 0:
        plt.plot(spectral_params, spectral_scores, marker='o', label='Spectral')
    if len(aff_params) > 0:
        plt.plot(aff_params, aff_scores, marker='s', label='Affinity Prop')
    
    plt.axhline(louvain_score, color='r', linestyle='--', label=f'Louvain (score={louvain_score:.2f})')
    plt.xlabel('Parameter value')
    plt.ylabel('Silhouette Score')
    plt.legend()
    plt.title(f'Clustering Methods Comparison (rho={rho})')
    
    # Visualize the best clustering
    all_scores = [s for s in spectral_scores if not np.isnan(s)] + \
                [s for s in aff_scores if not np.isnan(s)] + \
                [louvain_score]
    
    if len(all_scores) > 0:
        best_score = max(all_scores)
        if best_score == louvain_score:
            best_labels = louvain_labels
        elif best_score in spectral_scores:
            best_n = spectral_params[np.argmax(spectral_scores)]
            clustering = SpectralClustering(n_clusters=best_n, affinity='precomputed')
            best_labels = clustering.fit_predict(np.asarray(X))
        else:
            best_damping = aff_params[np.argmax(aff_scores)]
            clustering = AffinityPropagation(affinity='precomputed',
                                           damping=best_damping,
                                           random_state=42)
            best_labels = clustering.fit_predict(np.asarray(X))
        
        plt.subplot(1, 2, 2)
        G = nx.from_numpy_array(X)
        nx.draw_spring(G, node_color=best_labels, cmap=plt.cm.tab20, 
                      with_labels=False, node_size=50)
        plt.title(f'Best Clustering (Score={best_score:.2f})')
    
    plt.tight_layout()
    plt.show()
```
